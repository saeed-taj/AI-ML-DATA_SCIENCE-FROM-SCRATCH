#### Data preprocessing : Convert string representation of list into actual python list 

```python
transactions = df["Items"].apply(ast.literal_eval).tolist()
```

#### Find unique items count across all transactions

```python
flattened_item = [item for sublist in transactions for item in sublist]

unique_items_count = len(set(flattened_item))
```

```python
print(f"\nNumber of Unique Items: {unique_items_count}")
```

#### One-hot encoding using TransactionEncoder for Apriori

```python
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)

df_encoded = pd.DataFrame(te_ary, columns = te.columns_)

print( f"\nEncoded Transaction Shape (Rows, Unique Items): {df_encoded.shape}")
```

### Frequent Itemset Mining (Section C Implementation)

```python
# here minimum support threshold
min_supp_thresh = 0.05

frequent_itemset = apriori(df_encoded, min_support = min_supp_thresh, use_columns = True )

```
